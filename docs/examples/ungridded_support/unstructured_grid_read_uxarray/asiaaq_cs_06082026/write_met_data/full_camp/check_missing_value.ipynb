```python
import pandas as pd

f = "/glade/campaign/acom/acom-weather/emmons/ASIAAQ_obs/DC8/asiaaq-mrg10_dc8_20240327_RA_20260509.ict"

with open(f) as fh:
    nheader = int(fh.readline().split(",")[0])

df = pd.read_csv(f, skiprows=nheader)

# Force numeric where possible
df = df.apply(pd.to_numeric, errors="coerce")

for mv in [-9999, -888888, -1]:
    print(f"\n===== {mv} =====")

    found = False
    for col in df.columns:
        n = (df[col] == mv).sum()
        if n > 0:
            found = True
            print(f"{col}: {n}")

    if not found:
        print("No occurrences found")
```
