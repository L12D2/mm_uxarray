```python
import pandas as pd
p = an.paired['airnow_cam-chem-se'].obj
df = p.to_dataframe().reset_index() if not isinstance(p, pd.DataFrame) else p
print("columns:", list(df.columns))
num = [c for c in df.columns if str(df[c].dtype).startswith(("float","int"))]
for c in num:
    print(f"  {c}: {df[c].notnull().sum()} non-null")

# minute-of-hour for obs vs model time stamps
ot = pd.to_datetime(df.loc[df['temperature'].notnull(), 'time'])
print("obs minutes:", sorted(ot.dt.minute.unique()))
mcol = [c for c in df.columns if c == 'T' or c.startswith('T_') or 'cam-chem' in c]
print("model col(s):", mcol)
for c in mcol:
    mt = pd.to_datetime(df.loc[df[c].notnull(), 'time'])
    print(f"  {c} minutes:", sorted(mt.dt.minute.unique()))
```
